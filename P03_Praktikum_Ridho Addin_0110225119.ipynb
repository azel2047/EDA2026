```python
# TODO D1 — susun tabel penandaan Anda sendiri

# Buat dictionary bernama PENANDAAN berisi pasangan:
#     'nama_kolom': ('kategori', 'tindakan')
# Tindakan yang tersedia: 'hapus', 'hash', 'generalisasi', 'pertahankan'
# Kolom yang tidak Anda daftarkan dianggap 'bukan data pribadi' + 'pertahankan'.
PENANDAAN = {
    # Pengenal langsung
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik_dummy': ('pengenal langsung', 'hash'),
    'no_hp': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hapus'),
    'alamat_jalan': ('pengenal langsung', 'hapus'),

    # Quasi-identifier
    'tanggal_lahir': ('quasi-identifier', 'generalisasi'),
    'kelurahan': ('quasi-identifier', 'generalisasi'),
    'kecamatan': ('quasi-identifier', 'generalisasi'),

    # Data spesifik
    'status_disabilitas': ('data spesifik', 'pertahankan'),
    'jumlah_anak': ('data spesifik', 'generalisasi'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'generalisasi')
}

#
# Lalu bangun DataFrame 'tanda' dengan kolom: nama_kolom, kategori, tindakan
data_penandaan = []
for col in df.columns:
    if col in PENANDAAN:
        kategori, tindakan = PENANDAAN[col]
    else:
        kategori, tindakan = 'bukan data pribadi', 'pertahankan'

    data_penandaan.append({
        'nama_kolom': col,
        'kategori': kategori,
        'tindakan': tindakan
    })

tanda = pd.DataFrame(data_penandaan)

# dan cetak jumlah kolom per kategori.
print("Jumlah kolom berdasarkan kategori data:")
print(tanda['kategori'].value_counts())

# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom
```

```python
# TODO E1 — HAPUS
kolom_dihapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()

# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
aman = df.drop(columns=kolom_dihapus, errors='ignore')

# Cetak berapa kolom yang dibuang dan berapa yang tersisa.
print(f"Jumlah kolom yang dibuang : {len(kolom_dihapus)}")
print(f"Jumlah kolom yang tersisa : {aman.shape[1]}")
```

```python
# TODO E2 — HASH
import hashlib
import pandas as pd
import numpy as np

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan

    teks_gabungan = str(garam) + str(nilai)

    hasil_hash = hashlib.sha256(teks_gabungan.encode('utf-8')).hexdigest()[:12]
    return hasil_hash

# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
kolom_dihash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()

for col in kolom_dihash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama
print(f"Kolom yang di-hash: {kolom_dihash}")
if len(kolom_dihash) > 0:
    display(aman[kolom_dihash].head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
nilai_a1 = "3171234567890001"
nilai_a2 = "3171234567890001"
nilai_b  = "3179876543210009"

# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
hash_a1 = samarkan(nilai_a1, GARAM)
hash_a2 = samarkan(nilai_a2, GARAM)

print(f"Hash 1: {hash_a1}")
print(f"Hash 2: {hash_a2}")
print(f"Identik? {hash_a1 == hash_a2}\n")

# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
hash_b = samarkan(nilai_b, GARAM)

print(f"Hash Beda: {hash_b}")
print(f"Berbeda dengan Hash 1? {hash_a1 != hash_b}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

1. Yang tidak terlindungi: Lapisan perlindungan tambahan (kriptografi ekstra) terhadap serangan brute-force (coba-coba) atau rainbow table/dictionary attack akan hilang. Karena peretas tahu persis kata kunci rahasia (salt) dan algoritma yang digunakan, mereka bisa dengan mudah men-generate ulang hash dari database tebakan NIK atau Nomor HP untuk dicocokkan dengan data yang bocor.
2. Yang masih terlindungi: Teks/nilai asli di dalam database tetap berbentuk hash tersamar dan tidak bisa langsung dibaca (irreversible). Hash bersifat satu arah, sehingga datanya tidak bisa dikembalikan ke bentuk teks asli tanpa melakukan pencocokan (guessing).

```python
# TODO E4 — GENERALISASI
import pandas as pd
import numpy as np

ACUAN = pd.Timestamp('2025-09-01')

aman['tanggal_lahir'] = pd.to_datetime(aman['tanggal_lahir'])

# a. Hitung usia peserta pada tanggal acuan 1 September 2025
# Fix: Calculate age in days and then divide by 365.25 for approximate years
aman['usia'] = ((ACUAN - aman['tanggal_lahir']).dt.days / 365.25).astype(int)

# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
batas_bins = [0, 25, 35, 45, 55, float('inf')]
label_bins = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(aman['usia'], bins=batas_bins, labels=label_bins, right=False)

# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'], errors='ignore')

# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title()

# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik
print("Distribusi kelompok_umur:")
print(aman['kelompok_umur'].value_counts())
print("\nJumlah nilai unik kecamatan:", aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
def cek_k(data, kolom):
    # Dikelompokkan berdasarkan kombinasi kolom dan hitung ukuran tiap grup
    grup_kombinasi = data.groupby(kolom, observed=True).size()

    # Cari nilai k terkecil
    k_terkecil = grup_kombinasi.min()

    # Hitung jumlah grup yang rawan (hanya berisi 1 orang)
    grup_1_orang = (grup_kombinasi == 1).sum()

    # Total kombinasi grup unik
    total_grup = len(grup_kombinasi)

    # Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup
    return pd.Series({
        'k (grup terkecil)': k_terkecil,
        'grup berisi 1 orang': grup_1_orang,
        'total grup': total_grup
    })
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
kolom_awal = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']

risiko_awal = cek_k(df, kolom_awal)

print("Risiko SEBELUM penanganan:")
print(risiko_awal)
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_Meskipun tidak memuat nama atau nomor telepon, kombinasi dari atribut-atribut tersebut (quasi-identifiers) seringkali cukup unik dan bertindak seperti sidik jari digital untuk satu individu. Jika penyerang menyilangkan data ini dengan basis data publik eksternal (seperti daftar pemilih), identitas asli individu tersebut dapat dengan mudah dibongkar (terjadi linkage attack)._

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
kolom_f3 = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']

risiko_sesudah = cek_k(aman, kolom_f3)

print("Risiko SESUDAH generalisasi:")
print(risiko_sesudah)
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.

# STRATEGI 1: Keluarkan 'pendidikan_terakhir' dari quasi-identifier
kolom_f4 = ['kecamatan', 'jenis_kelamin', 'kelompok_umur']
risiko_f4_tahap1 = cek_k(aman, kolom_f4)
print("Risiko Tahap 1 (Tanpa pendidikan_terakhir):")
print(risiko_f4_tahap1)

# STRATEGI 2 (Tambahan jika tahap 1 masih memiliki k < 5):
# Gabungkan kecamatan yang jumlah pesertanya sangat sedikit menjadi 'Lainnya'
frekuensi_kecamatan = aman['kecamatan'].value_counts()
# Tentukan batas, misalnya kecamatan dengan peserta < 15 digabung
kecamatan_minor = frekuensi_kecamatan[frekuensi_kecamatan < 15].index
aman['kecamatan_gen'] = aman['kecamatan'].replace(kecamatan_minor, 'Lainnya')

kolom_f4_final = ['kecamatan_gen', 'jenis_kelamin', 'kelompok_umur']
risiko_f4_final = cek_k(aman, kolom_f4_final)

print("\nRisiko Final (Kecamatan digeneralisasi ke 'Lainnya'):")
print(risiko_f4_final)
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

_Kombinasi yang saya pilih adalah ['kecamatan', 'jenis_kelamin', 'kelompok_umur'] dengan tambahan modifikasi mengelompokkan wilayah kecamatan kecil menjadi "Lainnya". Dengan langkah ini, nilai target k $\ge$ 5 dapat dicapai.
