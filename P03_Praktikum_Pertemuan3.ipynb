## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
kolom_hapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()
aman = df.drop(columns=kolom_hapus).copy()

print(f"Kolom yang dibuang: {len(kolom_hapus)}")
print(f"Kolom yang tersisa: {aman.shape[1]}")
```

```python
# TODO E2 — HASH
def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    gabungan = str(garam) + str(nilai)
    return hashlib.sha256(gabungan.encode('utf-8')).hexdigest()[:12]

GARAM = 'praktikum-eda-kunci-rahasia-2025'

kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()
for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

print(aman[kolom_hash].head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
hash_a = samarkan('PESERTA_A', GARAM)
hash_b = samarkan('PESERTA_A', GARAM)
hash_c = samarkan('PESERTA_B', GARAM)

print(f"Hash A: {hash_a} | Hash B: {hash_b} | Identik? {hash_a == hash_b}")
print(f"Hash C: {hash_c} | Berbeda? {hash_a != hash_c}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

_(Nilai data asli tetap aman dari pembacaan langsung secara kasat mata. Namun, perlindungan terhadap serangan brute-force atau rainbow table akan runtuh karena penyerang dapat menghitung hash dari kamus nilai yang mungkin dan mencocokkannya secara langsung.)_

```python
# TODO E4 — GENERALISASI
ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta
usia = (ACUAN - df['tanggal_lahir']).dt.days // 365.25

# b. Buat kelompok umur
kelompok_umur = pd.cut(
    usia,
    bins=[0, 24, 34, 44, 54, np.inf],
    labels=['<25', '25-34', '35-44', '45-54', '55+']
)
aman['kelompok_umur'] = kelompok_umur

# c. Buang kolom tanggal_lahir dan kelurahan
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'], errors='ignore')

# d. Rapikan kolom kecamatan
aman['kecamatan'] = aman['kecamatan'].str.title().str.strip()

print("Kelompok umur terbanyak:", aman['kelompok_umur'].value_counts().idxmax())
print("Jumlah kecamatan unik:", aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
def cek_k(data, kolom):
    grup = data.groupby(kolom).size()
    k_val = grup.min()
    grup_1_orang = (grup == 1).sum()
    total_grup = len(grup)
    return pd.Series([k_val, grup_1_orang, total_grup], index=['k', 'grup_1_orang', 'total_grup'])
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
kolom_qi_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
print(cek_k(df, kolom_qi_sebelum))
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_(Kombinasi spesifik dari empat data tersebut membentuk fingerprint unik untuk setiap individu. Seseorang dapat dengan mudah mencocokkan tanggal lahir, alamat kelurahan, dan jenjang pendidikan ini dengan basis data publik lain (seperti DPT Pemilu atau media sosial) untuk mengungkap identitas asli peserta.)_

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
kolom_qi_sesudah = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
print(cek_k(aman, kolom_qi_sesudah))
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Langkah taktis: Kita keluarkan kolom 'pendidikan_terakhir' dari analisis atau kita buat kelompok umur lebih lebar
aman_super = aman.copy()
aman_super['kelompok_umur_lebar'] = pd.cut(
    (ACUAN - df['tanggal_lahir']).dt.days // 365.25,
    bins=[0, 34, 49, np.inf],
    labels=['<35', '35-49', '50+']
)

# Memakai kombinasi: ['kecamatan', 'jenis_kelamin', 'kelompok_umur_lebar']
kolom_qi_tantangan = ['kecamatan', 'jenis_kelamin', 'kelompok_umur_lebar']
res = cek_k(aman_super, kolom_qi_tantangan)
print(res)
# Jika k >= 5 tercapai, maka target terpenuhi
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

_(Kombinasi yang dipilih: kecamatan, jenis_kelamin, pendidikan_grup (digeneralisasi menjadi 2 jenjang: Dasar/Menengah dan Tinggi), serta umur_lebar (3 kelompok: <35, 35–49, 50+).Nilai $k$ yang dicapai: $\ge 5$ (tidak ada grup dengan jumlah peserta di bawah 5).Informasi yang dikorbankan: Granularitas tingkatan pendidikan detail (perbedaan D3 vs S1/S2 hilang) dan presisi kelompok umur 10-tahunan.)_

```python
# TODO F5 — sel kecil pada tabel agregat
# a. Buat tabel silang jumlah peserta menurut kecamatan x kategori_usaha
tabel_silang = pd.crosstab(aman['kecamatan'], aman['kategori_usaha'])

# b. Hitung berapa sel yang terisi tetapi berisi kurang dari 5 peserta
sel_terisi = tabel_silang[tabel_silang > 0]
sel_kecil = (sel_terisi < 5).sum().sum()

print("Tabel Silang (Kecamatan x Kategori Usaha):")
print(tabel_silang.head())
print(f"\nJumlah sel terisi yang berisi < 5 peserta: {sel_kecil}")
```

```python
# G4 — jalankan penanganan pada data tim dan periksa nilai k
# Salin pola Bagian E dan F, ganti dataset dan nama kolomnya.
# Simpan hasilnya ke data/processed/ dan JANGAN unggah data/raw/

import os
import hashlib
import pandas as pd
import numpy as np

# 1. Pastikan struktur folder tersedia
os.makedirs('data/processed', exist_ok=True)

# 2. Muat dataset mentah
# Ganti nama berkas jika menggunakan berkas data tim Anda sendiri
FILE_INPUT = 'peserta_pelatihan_umkm_2025.csv'
df_tim = pd.read_csv(FILE_INPUT)

# 3. Hapus Direct Identifiers
kolom_hapus_tim = [
    'nama_lengkap', 'nik_dummy', 'no_hp', 'email',
    'alamat_jalan', 'kelurahan', 'penghasilan_pribadi_bulanan'
]
df_bersih = df_tim.drop(columns=[col for col in kolom_hapus_tim if col in df_tim.columns]).copy()

# 4. Pseudonimisasi ID Peserta (Gunakan salt dari environment / rahasia)
GARAM_TIM = os.getenv('SALT_KEY', 'rahasia_proyek_eda_tim_2025')

def hash_id(val, salt):
    if pd.isna(val):
        return np.nan
    return hashlib.sha256(f"{salt}{val}".encode('utf-8')).hexdigest()[:12]

if 'id_peserta' in df_bersih.columns:
    df_bersih['id_peserta'] = df_bersih['id_peserta'].apply(lambda x: hash_id(x, GARAM_TIM))

# 5. Generalisasi: Kelompok Umur & Wilayah
if 'tanggal_lahir' in df_bersih.columns:
    df_bersih['tanggal_lahir'] = pd.to_datetime(df_bersih['tanggal_lahir'])
    ACUAN = pd.Timestamp('2025-09-01')
    usia = (ACUAN - df_bersih['tanggal_lahir']).dt.days // 365.25
    df_bersih['kelompok_umur'] = pd.cut(
        usia,
        bins=[0, 34, 49, np.inf],
        labels=['<35', '35-49', '50+']
    )
    df_bersih.drop(columns=['tanggal_lahir'], inplace=True)

if 'kecamatan' in df_bersih.columns:
    df_bersih['kecamatan'] = df_bersih['kecamatan'].astype(str).str.strip().str.title()

# 6. Generalisasi: Pendidikan Terakhir
map_pend = {
    'SD': 'Dasar/Menengah', 'SMP': 'Dasar/Menengah', 'SMA/SMK': 'Dasar/Menengah',
    'D3': 'Tinggi', 'S1': 'Tinggi', 'S2': 'Tinggi'
}
if 'pendidikan_terakhir' in df_bersih.columns:
    df_bersih['pendidikan_terakhir'] = df_bersih['pendidikan_terakhir'].map(map_pend).fillna('Lainnya')

# 7. Uji Sisa Risiko (k-anonymity)
def evaluasi_k(data, qi_cols):
    grup = data.groupby(qi_cols, observed=True).size()
    return pd.Series({
        'k_minimum': grup.min(),
        'grup_1_orang': (grup == 1).sum(),
        'grup_dibawah_5': (grup < 5).sum(),
        'total_kombinasi': len(grup)
    })

qi_eval = [c for c in ['kecamatan', 'jenis_kelamin', 'kelompok_umur', 'pendidikan_terakhir'] if c in df_bersih.columns]
hasil_eval = evaluasi_k(df_bersih, qi_eval)

print("--- Hasil Pengukuran K-Anonymity Data Tim ---")
print(hasil_eval)

# 8. Simpan hanya ke data/processed/ (JANGAN simpan data mentah)
OUTPUT_FILE = 'data/processed/dataset_tim_bersih.csv'
df_bersih.to_csv(OUTPUT_FILE, index=False)
print(f"\nDataset yang aman berhasil disimpan di: {OUTPUT_FILE}")
```
