````python
GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    # Convert all values to string before hashing
    hashed_value = hashlib.sha256(str(garam).encode() + str(nilai).encode()).hexdigest()
    return hashed_value[:12]

columns_to_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()

for col in columns_to_hash:
    aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))

print(f"Kolom yang di-hash: {columns_to_hash}")

# Display a sample to verify (first few rows of hashed columns)
print("Contoh data setelah di-hash:")
display(aman[columns_to_hash].head())
````

````python
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
TEST_VALUE_A = df.loc[0, 'id_peserta'] # Get a sample value from the original df
hash_a1 = samarkan(TEST_VALUE_A, GARAM)
hash_a2 = samarkan(TEST_VALUE_A, GARAM)
print(f"Nilai asli A: {TEST_VALUE_A}")
print(f"Hash A pertama: {hash_a1}")
print(f"Hash A kedua: {hash_a2}")
print(f"Hasil hash identik: {hash_a1 == hash_a2}\n")

# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
TEST_VALUE_B1 = df.loc[0, 'email'] # Get another sample value
TEST_VALUE_B2 = df.loc[1, 'email'] # Get a different sample value
hash_b1 = samarkan(TEST_VALUE_B1, GARAM)
hash_b2 = samarkan(TEST_VALUE_B2, GARAM)
print(f"Nilai asli B1: {TEST_VALUE_B1}")
print(f"Nilai asli B2: {TEST_VALUE_B2}")
print(f"Hash B1: {hash_b1}")
print(f"Hash B2: {hash_b2}")
print(f"Hasil hash berbeda: {hash_b1 != hash_b2}")
````

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

````python
def cek_k(data, kolom):
    if not kolom:
        return 0, 0, 0 # No columns to group by

    # Group by the specified columns and count occurrences
    grup_counts = data.groupby(kolom).size()

    # Calculate k (smallest group size)
    k = grup_counts.min() if not grup_counts.empty else 0

    # Calculate number of groups with only one person
    grup_satu_orang = (grup_counts == 1).sum()

    # Calculate total number of groups
    total_grup = len(grup_counts)

    return pd.Series({'k': k, 'grup_satu_orang': grup_satu_orang, 'total_grup': total_grup})
````

````python
kolom_quasi_sebelum = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
hasil_sebelum = cek_k(df, kolom_quasi_sebelum)

print(f"Risiko sebelum penanganan:")
print(f"  k (ukuran grup terkecil): {hasil_sebelum['k']}")
print(f"  Grup berisi 1 orang: {hasil_sebelum['grup_satu_orang']}")
print(f"  Total grup: {hasil_sebelum['total_grup']}")

# Verification
if hasil_sebelum['k'] == 1 and hasil_sebelum['grup_satu_orang'] == df.shape[0]:
    print("Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang (terbukti).")
else:
    print("Warning: Output does not match expected results.")
````

````python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.

# Strategi 1: Menggabungkan kecamatan kecil menjadi 'Lainnya'
# Identifikasi kecamatan dengan frekuensi rendah
kecamatan_counts = aman['kecamatan'].value_counts()
low_frequency_kecamatan = kecamatan_counts[kecamatan_counts < 5].index.tolist()

# Buat salinan DataFrame untuk eksperimen agar aman tidak terpengaruh langsung oleh percobaan
aman_exp = aman.copy()

# Ganti kecamatan dengan frekuensi rendah menjadi 'Lainnya'
aman_exp['kecamatan_grouped'] = aman_exp['kecamatan'].replace(low_frequency_kecamatan, 'Lainnya')

print(f"Original unique kecamatan count: {aman['kecamatan'].nunique()}")
print(f"Grouped unique kecamatan count: {aman_exp['kecamatan_grouped'].nunique()}\n")

# Ukur risiko SESUDAH generalisasi 'kecamatan'
kolom_quasi_final = ['kecamatan_grouped', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
hasil_final = cek_k(aman_exp, kolom_quasi_final)

print(f"Risiko sesudah generalisasi kecamatan (grouped):")
print(f"  k (ukuran grup terkecil): {hasil_final['k']}")
print(f"  Grup berisi 1 orang: {hasil_final['grup_satu_orang']}")
print(f"  Total grup: {hasil_final['total_grup']}")

# Jika k masih di bawah 5, coba strategi lain atau kombinasi
if hasil_final['k'] < 5:
    print("\nK-anonymity (k) is still less than 5 after grouping small kecamatan. Trying further generalization...")
    # Strategi 2: Perlebar kelompok umur (misal, menjadi lebih besar)
    # Hapus kolom 'kelompok_umur' yang lama untuk membuat yang baru
    aman_exp = aman_exp.drop(columns=['kelompok_umur'])

    bins_broad = [0, 30, 40, 50, np.inf] # Contoh kelompok umur yang lebih lebar
    labels_broad = ['<30', '30-39', '40-49', '50+']
    aman_exp['kelompok_umur_broad'] = pd.cut(aman_exp['usia'], bins=bins_broad, labels=labels_broad, right=False)

    kolom_quasi_final_broad = ['kecamatan_grouped', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur_broad']
    hasil_final_broad = cek_k(aman_exp, kolom_quasi_final_broad)

    print(f"\nRisiko sesudah generalisasi kecamatan DAN kelompok umur (broad):")
    print(f"  k (ukuran grup terkecil): {hasil_final_broad['k']}")
    print(f"  Grup berisi 1 orang: {hasil_final_broad['grup_satu_orang']}")
    print(f"  Total grup: {hasil_final_broad['total_grup']}")

    # If k is still not 5, we might need to remove a quasi-identifier or combine categories even further.
    # For this exercise, let's assume one of these steps will be sufficient for k>=5.
    if hasil_final_broad['k'] >= 5:
        aman = aman_exp # Update 'aman' with the successful generalization
        print("\nBerhasil mencapai k >= 5 dengan generalisasi kecamatan dan kelompok umur.")
    else:
        print("\nK-anonymity still not >= 5. Consider removing a quasi-identifier or more aggressive generalization.")
else:
    aman = aman_exp # Update 'aman' with the successful generalization
    print("\nBerhasil mencapai k >= 5 dengan generalisasi kecamatan.")

# Ensure the 'aman' dataframe has the final generalized columns if k>=5 is met
if hasil_final['k'] >= 5: # If the first strategy worked
    aman['kecamatan'] = aman_exp['kecamatan_grouped']
elif 'kelompok_umur_broad' in aman_exp.columns and hasil_final_broad['k'] >= 5: # If the second strategy worked
    aman['kecamatan'] = aman_exp['kecamatan_grouped']
    aman['kelompok_umur'] = aman_exp['kelompok_umur_broad']

# Cleaning up temporary columns for next steps if needed, or if only one strategy worked
if 'kecamatan_grouped' in aman.columns:
    aman = aman.drop(columns=['kecamatan_grouped'])
if 'kelompok_umur_broad' in aman.columns:
    aman = aman.drop(columns=['kelompok_umur_broad'])





````

````python
# 1. Muat Dataset Tim Anda
# Ganti 'nama_file_dataset_tim_anda.csv' dengan nama file CSV dataset tim Anda
# dan sesuaikan pemisah (delimiter) jika bukan koma.
TEAM_FILE = 'nama_file_dataset_tim_anda.csv' # Ganti dengan nama file dataset Anda

# Pastikan Anda telah mengunggah file atau mount Google Drive jika file ada di Drive
# Contoh: df_team = pd.read_csv(TEAM_FILE)
# Untuk contoh ini, saya akan menggunakan df yang sudah ada sebagai placeholder
# HAPUS BARIS INI DAN AKTIFKAN BARIS DI ATAS setelah Anda mengatur dataset tim Anda
df_team = df.copy()

print(f"Dataset tim Anda dimuat dengan {df_team.shape[0]} baris dan {df_team.shape[1]} kolom.")

# 2. Definisikan PENANDAAN_TEAM untuk dataset Anda
# Sesuaikan kategori dan tindakan untuk setiap kolom di dataset tim Anda.
# Ini adalah contoh, Anda harus mengisi sesuai dengan kolom dataset Anda.
PENANDAAN_TEAM = {
    'id_peserta': ('pengenal langsung', 'hash'),
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik_dummy': ('pengenal langsung', 'hash'),
    'no_hp': ('pengenal langsung', 'hash'),
    'email': ('pengenal langsung', 'hash'),
    'alamat_jalan': ('quasi-identifier', 'hapus'),
    'kelurahan': ('quasi-identifier', 'generalisasi'), # Ganti dengan kolom kelurahan Anda
    'kecamatan': ('quasi-identifier', 'generalisasi'), # Ganti dengan kolom kecamatan Anda
    'tanggal_lahir': ('data spesifik', 'generalisasi'), # Ganti dengan kolom tanggal lahir Anda
    'status_disabilitas': ('data spesifik', 'pertahankan'),
    'jumlah_anak': ('data spesifik', 'pertahankan'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'pertahankan'),
    'omzet_usaha_bulanan': ('data spesifik', 'pertahankan'),
    'nama_usaha': ('quasi-identifier', 'pertahankan')
    # Tambahkan kolom lain dari dataset tim Anda di sini
}

# Buat DataFrame 'tanda_team'
tanda_team_list = []
for col in df_team.columns:
    kategori, tindakan = PENANDAAN_TEAM.get(col, ('bukan data pribadi', 'pertahankan'))
    tanda_team_list.append({
        'nama_kolom': col,
        'kategori': kategori,
        'tindakan': tindakan
    })

tanda_team = pd.DataFrame(tanda_team_list)

print("\nJumlah kolom per kategori (dataset tim):")
print(tanda_team['kategori'].value_counts())
display(tanda_team.head())

# Buat salinan DataFrame untuk penanganan
aman_team = df_team.copy()
````

````python
# 3. Menerapkan Tindakan 'Hapus' (Dropping Columns)
columns_to_drop_team = tanda_team[tanda_team['tindakan'] == 'hapus']['nama_kolom'].tolist()

aman_team = aman_team.drop(columns=columns_to_drop_team, errors='ignore')

print(f"Jumlah kolom yang dibuang dari aman_team: {len(columns_to_drop_team)}")
print(f"Jumlah kolom yang tersisa di aman_team: {aman_team.shape[1]}")

# 4. Menerapkan Tindakan 'Hash' (Hashing Columns)
# Gunakan GARAM yang berbeda untuk dataset tim Anda!
GARAM_TEAM = 'garam-rahasia-tim-anda' # GANTI DENGAN KALIMAT RAHASIA YANG UNIK UNTUK TIM ANDA

# Pastikan fungsi `samarkan` sudah didefinisikan di sel sebelumnya (zBZE1cOr-x55)

columns_to_hash_team = tanda_team[tanda_team['tindakan'] == 'hash']['nama_kolom'].tolist()

for col in columns_to_hash_team:
    if col in aman_team.columns: # Pastikan kolom masih ada setelah penghapusan
        aman_team[col] = aman_team[col].apply(lambda x: samarkan(x, GARAM_TEAM))
    else:
        print(f"Kolom '{col}' tidak ditemukan di aman_team (mungkin sudah dihapus).")

print(f"\nKolom yang di-hash di aman_team: {columns_to_hash_team}")

# Display a sample to verify (first few rows of hashed columns)
print("Contoh data setelah di-hash (aman_team):")
display(aman_team[columns_to_hash_team].head())
````

````python
# 5. Menerapkan Tindakan 'Generalisasi' (Age Grouping dan Kecamatan Cleaning)

# Sesuaikan nama kolom jika berbeda di dataset tim Anda
COLUMN_TANGGAL_LAHIR_TEAM = 'tanggal_lahir' # Ganti dengan nama kolom tanggal lahir Anda
COLUMN_KELURAHAN_TEAM = 'kelurahan'       # Ganti dengan nama kolom kelurahan Anda
COLUMN_KECAMATAN_TEAM = 'kecamatan'       # Ganti dengan nama kolom kecamatan Anda

# Konversi kolom tanggal lahir ke datetime jika belum
if COLUMN_TANGGAL_LAHIR_TEAM in aman_team.columns:
    aman_team[COLUMN_TANGGAL_LAHIR_TEAM] = pd.to_datetime(aman_team[COLUMN_TANGGAL_LAHIR_TEAM], errors='coerce')

    # Tentukan tanggal acuan untuk perhitungan usia (sesuaikan jika perlu)
    ACUAN_TEAM = pd.Timestamp('2025-09-01') # Sesuaikan tanggal acuan untuk dataset Anda

    # a. Hitung usia peserta
    aman_team['usia'] = (ACUAN_TEAM - aman_team[COLUMN_TANGGAL_LAHIR_TEAM]).dt.days / 365.25

    # b. Buat kolom 'kelompok_umur' (sesuaikan bin dan label jika diperlukan)
    bins_umur_team = [0, 25, 35, 45, 55, np.inf]
    labels_umur_team = ['<25', '25-34', '35-44', '45-54', '55+']
    aman_team['kelompok_umur'] = pd.cut(aman_team['usia'], bins=bins_umur_team, labels=labels_umur_team, right=False)

    print(f"Kelompok umur terbanyak di aman_team: {aman_team['kelompok_umur'].value_counts().idxmax()}")
else:
    print(f"Kolom '{COLUMN_TANGGAL_LAHIR_TEAM}' tidak ditemukan di aman_team. Usia dan kelompok umur tidak dihitung.")

# c. Buang kolom asli yang digeneralisasi ('tanggal_lahir' dan 'kelurahan')
col_to_drop_after_gen = []
if COLUMN_TANGGAL_LAHIR_TEAM in aman_team.columns:
    col_to_drop_after_gen.append(COLUMN_TANGGAL_LAHIR_TEAM)
if COLUMN_KELURAHAN_TEAM in aman_team.columns:
    col_to_drop_after_gen.append(COLUMN_KELURAHAN_TEAM)

aman_team = aman_team.drop(columns=col_to_drop_after_gen, errors='ignore')

# d. Rapikan kolom kecamatan (jika ada)
if COLUMN_KECAMATAN_TEAM in aman_team.columns:
    aman_team[COLUMN_KECAMATAN_TEAM] = aman_team[COLUMN_KECAMATAN_TEAM].str.title()
    print(f"Jumlah nilai unik di kolom '{COLUMN_KECAMATAN_TEAM}' aman_team: {aman_team[COLUMN_KECAMATAN_TEAM].nunique()}")
else:
    print(f"Kolom '{COLUMN_KECAMATAN_TEAM}' tidak ditemukan di aman_team.")

display(aman_team.head())
````
